```python

room_a = input("Enter Room A status (D for Dirty, C for Clean): ").upper()
room_b = input("Enter Room B status (D for Dirty, C for Clean): ").upper()
position = input("Enter vacuum position (A/B): ").upper()

while room_a == "D" or room_b == "D":

    if position == "A":

        if room_a == "D":
            print("\nVacuum is in Room A.")
            print("Room A is dirty.")
            print("Action: SUCK")
            room_a = "C"
            print("Room A is now clean.")

        else:
            print("\nRoom A is already clean.")
            print("Action: MOVE RIGHT")
            position = "B"


    else:

        if room_b == "D":
            print("\nVacuum is in Room B.")
            print("Room B is dirty.")
            print("Action: SUCK")
            room_b = "C"
            print("Room B is now clean.")

        else:
            print("\nRoom B is already clean.")
            print("Action: MOVE LEFT")
            position = "A"



print("\nBoth rooms are clean.")
print("Cleaning completed.")
```

```python
import random

a = [[" ", " ", " "],
     [" ", " ", " "],
     [" ", " ", " "]]

user = input("Choose X or O: ").upper()

if user == "X":
    computer = "O"
else:
    computer = "X"

def display():
    print()
    print(a[0][0], "|", a[0][1], "|", a[0][2])
    print("--+---+--")
    print(a[1][0], "|", a[1][1], "|", a[1][2])
    print("--+---+--")
    print(a[2][0], "|", a[2][1], "|", a[2][2])
    print()

def win(symbol):

    if a[0][0] == symbol and a[0][1] == symbol and a[0][2] == symbol:
        return True

    if a[1][0] == symbol and a[1][1] == symbol and a[1][2] == symbol:
        return True

    if a[2][0] == symbol and a[2][1] == symbol and a[2][2] == symbol:
        return True

    if a[0][0] == symbol and a[1][0] == symbol and a[2][0] == symbol:
        return True

    if a[0][1] == symbol and a[1][1] == symbol and a[2][1] == symbol:
        return True

    if a[0][2] == symbol and a[1][2] == symbol and a[2][2] == symbol:
        return True

    if a[0][0] == symbol and a[1][1] == symbol and a[2][2] == symbol:
        return True

    if a[0][2] == symbol and a[1][1] == symbol and a[2][0] == symbol:
        return True

    return False


while True:

    position = int(input("Enter position (1-9): "))
    row = (position - 1) // 3
    column = (position - 1) % 3

    if position >= 1 and position <= 9 and a[row][column] == " ":
        a[row][column] = user
    else:
        print("Invalid position")
        continue

    display()

    if win(user):
        print(user, "wins!")
        break

    full = True

    for i in range(3):
        for j in range(3):
            if a[i][j] == " ":
                full = False

    if full:
        print("DRAW")
        break

    while True:

        computer_position = random.randint(1, 9)
        row = (computer_position - 1) // 3
        column = (computer_position - 1) % 3

        if a[row][column] == " ":
            a[row][column] = computer
            break

    print("Computer chose:", computer_position)

    display()

    if win(computer):
        print(computer, "wins!")
        break

    full = True

    for i in range(3):
        for j in range(3):
            if a[i][j] == " ":
                full = False

    if full:
        print("DRAW")
        break
```
