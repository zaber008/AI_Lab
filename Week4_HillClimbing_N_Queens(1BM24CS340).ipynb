```python
import itertools

def calculate_cost(state):
    conflicts = 0
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            if abs(i - j) == abs(state[i] - state[j]):
                conflicts += 1
    return conflicts

def get_best_neighbor(state):
    best_neighbor = None
    best_cost = float('inf')
    n = len(state)

    for i, j in itertools.combinations(range(n), 2):
        neighbor = list(state)
        neighbor[i], neighbor[j] = neighbor[j], neighbor[i]

        cost = calculate_cost(neighbor)

        if cost < best_cost:
            best_cost = cost
            best_neighbor = neighbor

    return best_neighbor, best_cost

def hill_climbing(initial_state):
    current_state = initial_state
    current_cost = calculate_cost(current_state)

    print(f"Initial State: {current_state} | Cost: {current_cost}")

    step = 1
    while True:
        if current_cost == 0:
            print("\nGoal state reached!")
            break

        neighbor_state, neighbor_cost = get_best_neighbor(current_state)

        if neighbor_cost >= current_cost:
            print("\nStuck in local maximum/flat local area. Terminating.")
            break

        print(f"Step {step} -> Moving to neighbor: {neighbor_state} | Cost: {neighbor_cost}")
        current_state = neighbor_state
        current_cost = neighbor_cost
        step += 1

    return current_state

initial_state = [3, 1, 2, 0]
final_state = hill_climbing(initial_state)
print(f"Final Configuration: {final_state}")
```
